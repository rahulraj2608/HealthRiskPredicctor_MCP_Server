```python
import numpy as np
from google.colab import files



# Option B: Load directly into variables
data = np.load('lstm_processed_sequences.npz')

X_train = data['X_train']  # Shape: (4055, 7, 29)
X_test  = data['X_test']   # Shape: (1014, 7, 29)
y_train = data['y_train']  # Shape: (4055, 3)
y_test  = data['y_test']   # Shape: (1014, 3)

print("Loaded X_train shape:", X_train.shape)
print("Loaded y_train shape:", y_train.shape)
```

# **LSTM**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout

# 1. Verify shapes of loaded sequence arrays
time_steps = X_train.shape[1]  # 7
num_features = X_train.shape[2] # 29
num_targets = y_train.shape[1]   # 3

# 2. Build Multi-Output LSTM Architecture
model = Sequential([
    LSTM(64, return_sequences=True, input_shape=(time_steps, num_features)),
    Dropout(0.2),
    LSTM(32, return_sequences=False),
    Dropout(0.2),
    Dense(16, activation='relu'),
    Dense(num_targets)  # Outputs 3 targets
])

model.compile(optimizer='adam', loss='mse', metrics=['mae'])

# 3. Train the model
print("Training LSTM Model...")
history = model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=50,
    batch_size=32,
    verbose=1
)

# 4. Generate Predictions on Test Set
y_pred = model.predict(X_test)
y_test_orig = y_test

# 5. Calculate and Display Performance Metrics Table
target_names = ['Respiratory Disease Rate', 'Cardio Mortality Rate', 'Heat-Related Admissions']
lstm_results = []

for i, target_name in enumerate(target_names):
    r2 = r2_score(y_test_orig[:, i], y_pred[:, i])
    mse = mean_squared_error(y_test_orig[:, i], y_pred[:, i])
    mae = mean_absolute_error(y_test_orig[:, i], y_pred[:, i])

    lstm_results.append({
        'Target Outcome': target_name,
        'LSTM R² Score': round(r2, 4),
        'LSTM MSE': round(mse, 4),
        'LSTM MAE': round(mae, 4)
    })

results_df = pd.DataFrame(lstm_results)
display(results_df.style.background_gradient(subset=['LSTM R² Score'], cmap='viridis'))

# 6. Plot Actual vs. Predicted Time-Series Graphs
fig, axes = plt.subplots(3, 1, figsize=(14, 10), sharex=True)
plot_samples = min(150, len(y_test_orig))

for i, target_name in enumerate(target_names):
    axes[i].plot(y_test_orig[:plot_samples, i], label='Actual', color='#1f77b4', linewidth=2)
    axes[i].plot(y_pred[:plot_samples, i], label='Predicted (LSTM)', color='#ff7f0e', linestyle='--', linewidth=2)

    axes[i].set_title(f'LSTM Predictions vs Actual: {target_name}', fontsize=12, fontweight='bold')
    axes[i].set_ylabel('Value', fontsize=10)
    axes[i].legend(loc='upper right')
    axes[i].grid(True, linestyle=':', alpha=0.6)

axes[-1].set_xlabel('Test Sample Index (Time Steps)', fontsize=11)
plt.tight_layout()
plt.show()
```

# **Transformers**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import tensorflow as tf
from tensorflow.keras import layers, models

# 1. Inspect Input Dimensions
time_steps = X_train.shape[1]   # 7
num_features = X_train.shape[2]  # 29
num_targets = y_train.shape[1]    # 3

# 2. Build Transformer Architecture for Time-Series
def build_transformer_model(time_steps, num_features, num_targets):
    inputs = layers.Input(shape=(time_steps, num_features))

    # Linear projection to model dimension (d_model)
    x = layers.Dense(64)(inputs)

    # Positional Embedding / Encoding
    positions = tf.range(start=0, limit=time_steps, delta=1)
    pos_encoding = layers.Embedding(input_dim=time_steps, output_dim=64)(positions)
    x = x + pos_encoding

    # --- Transformer Encoder Block 1 ---
    attention_output1 = layers.MultiHeadAttention(num_heads=4, key_dim=16)(x, x)
    x1 = layers.Add()([x, attention_output1])
    x1 = layers.LayerNormalization(epsilon=1e-6)(x1)

    feed_forward1 = models.Sequential([
        layers.Dense(128, activation="relu"),
        layers.Dense(64),
    ])(x1)
    x1 = layers.Add()([x1, feed_forward1])
    x1 = layers.LayerNormalization(epsilon=1e-6)(x1)

    # --- Transformer Encoder Block 2 ---
    attention_output2 = layers.MultiHeadAttention(num_heads=4, key_dim=16)(x1, x1)
    x2 = layers.Add()([x1, attention_output2])
    x2 = layers.LayerNormalization(epsilon=1e-6)(x2)

    feed_forward2 = models.Sequential([
        layers.Dense(128, activation="relu"),
        layers.Dense(64),
    ])(x2)
    x2 = layers.Add()([x2, feed_forward2])
    x2 = layers.LayerNormalization(epsilon=1e-6)(x2)

    # Global Pooling & Output Dense Layers
    x_pooled = layers.GlobalAveragePooling1D()(x2)
    x_pooled = layers.Dropout(0.2)(x_pooled)
    x_pooled = layers.Dense(32, activation="relu")(x_pooled)

    outputs = layers.Dense(num_targets)(x_pooled)

    return models.Model(inputs=inputs, outputs=outputs, name="Time_Series_Transformer")

transformer_model = build_transformer_model(time_steps, num_features, num_targets)
transformer_model.compile(optimizer='adam', loss='mse', metrics=['mae'])

transformer_model.summary()

# 3. Train Transformer Model
print("\nTraining Time-Series Transformer Model...")
history_tf = transformer_model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=50,
    batch_size=32,
    verbose=1
)

# 4. Generate Predictions on Test Set
y_pred_tf = transformer_model.predict(X_test)
y_test_orig = y_test

# 5. Compute Metrics Table
target_names = ['Respiratory Disease Rate', 'Cardio Mortality Rate', 'Heat-Related Admissions']
tf_results = []

for i, target_name in enumerate(target_names):
    r2 = r2_score(y_test_orig[:, i], y_pred_tf[:, i])
    mse = mean_squared_error(y_test_orig[:, i], y_pred_tf[:, i])
    mae = mean_absolute_error(y_test_orig[:, i], y_pred_tf[:, i])

    tf_results.append({
        'Target Outcome': target_name,
        'Transformer R² Score': round(r2, 4),
        'Transformer MSE': round(mse, 4),
        'Transformer MAE': round(mae, 4)
    })

tf_results_df = pd.DataFrame(tf_results)
display(tf_results_df.style.background_gradient(subset=['Transformer R² Score'], cmap='viridis'))

# 6. Plot Actual vs Predicted Graphs
fig, axes = plt.subplots(3, 1, figsize=(14, 10), sharex=True)
plot_samples = min(150, len(y_test_orig))

for i, target_name in enumerate(target_names):
    axes[i].plot(y_test_orig[:plot_samples, i], label='Actual', color='#1f77b4', linewidth=2)
    axes[i].plot(y_pred_tf[:plot_samples, i], label='Predicted (Transformer)', color='#2ca02c', linestyle='--', linewidth=2)

    axes[i].set_title(f'Transformer Predictions vs Actual: {target_name}', fontsize=12, fontweight='bold')
    axes[i].set_ylabel('Value', fontsize=10)
    axes[i].legend(loc='upper right')
    axes[i].grid(True, linestyle=':', alpha=0.6)

axes[-1].set_xlabel('Test Sample Index (Time Steps)', fontsize=11)
plt.tight_layout()
plt.show()
```
